```python
prompts = """
prompts:
  - task: check_document_query

    content: |
      Determine whether the user's question is about the uploaded documents.

      If YES:
        return ALLOW

      If NO:
        return BLOCK

  - task: answer_from_documents

    content: |
      Answer ONLY using the uploaded document.

      If the answer is not present,
      reply:

      "I'm only able to answer questions based on the uploaded documents."
"""

with open("/content/drive/MyDrive/Project folder/config/prompts.yml","w") as f:
    f.write(prompts)

print("prompts.yml created")
```

**Since we don't yet have an LLM connected, we'll simulate the behavior.**

```python
document_keywords = [
    "machine learning",
    "deep learning",
    "supervised",
    "unsupervised",
    "neural network",
    "regression",
    "classification",
    "pdf",
    "tutorial"
]
```

```python
def guardrails(user_query):

    query = user_query.lower()

    for keyword in document_keywords:

        if keyword in query:
            return "ALLOW"

    return "BLOCK"
```

```python
questions = [

    "Explain supervised learning",

    "What is deep learning?",

    "Who is the President of India?",

    "What is IPL?",

    "Summarize the neural network chapter",

    "Tell me today's weather"

]

for q in questions:

    result = guardrails(q)

    print("="*50)
    print("Question:", q)

    if result=="ALLOW":
        print("✅ Allowed")

    else:
        print("❌ Blocked")
```

```python
def chatbot(query):

    result = guardrails(query)

    if result=="BLOCK":

        return "I'm only able to answer questions based on the uploaded documents."

    return "Searching uploaded documents..."
```

```python
while True:

    query = input("User: ")

    if query.lower()=="exit":
        break

    print(chatbot(query))
```
