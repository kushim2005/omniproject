## 2. API End-to-End Integration Testing
We use `requests` to test the full pipeline from document upload to agentic querying.

````python
import requests
import time

API_BASE_URL = "http://localhost:8000"

def test_health():
    print("Testing API Health...")
    try:
        res = requests.get(f"{API_BASE_URL}/health")
        if res.status_code == 200:
            print("✅ Health Check Passed:", res.json())
        else:
            print("❌ Health Check Failed:", res.status_code)
    except Exception as e:
        print("❌ Connection Error:", e)

test_health()
````

````python
def test_agent_query():
    print("Testing LangGraph Agent Routing...")
    payload = {
        "query": "Find charts showing the company revenue distribution",
        "top_k": 3,
        "search_text": True,
        "search_images": True
    }
    try:
        res = requests.post(f"{API_BASE_URL}/query", json=payload)
        if res.status_code == 200:
            data = res.json()
            print("✅ Query Success!")
            print(f"-> Text Results Found: {len(data.get('text_results', []))}")
            print(f"-> Image Results Found: {len(data.get('image_results', []))}")
        else:
            print("❌ Query Failed:", res.text)
    except Exception as e:
        print("❌ Connection Error:", e)

# Uncomment to run test if API is active
# test_agent_query()
````
